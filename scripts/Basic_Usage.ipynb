```python
mygrid = lcpom.Grid((20,20,20),(10,10,10))
```

```python
coords = lcpom.lcsystems.make_grid(mygrid.info)
orderparameter = np.ones(mygrid.info.shape)*0.58
orderfield = np.random.rand(mygrid.info.size,3)
interface = np.ones(mygrid.info.size)
normal = np.zeros((mygrid.info.size,3))
normal[:,2] = np.ones(mygrid.info.size)
```

```python
mylcgrid = lcpom.LCGrid(mygrid,orderparameter,orderfield,interface,normal)
Rp = lcpom.utils.tools.rotation_matrix(np.deg2rad(90))
```
